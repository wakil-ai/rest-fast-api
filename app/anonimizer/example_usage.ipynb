```python
from encryptor import UzbekRussianPIIDetector

detector = UzbekRussianPIIDetector()
```

```python
original_text = """
    Договор №123-2024 от 15.03.2024
    
    Продавец: ООО "Технологии Будущего", СТИР 123456789
    Директор: Иванов Петр Сергеевич
    Телефон: +998 90 123 45 67
    Email: ivanov@example.uz
    Р/с: 12345678901234567890, МФО: 00014
    
    Покупатель: Oybek Eraliev
    ЖШШИР: 12345678901234
    Паспорт: AA1234567
    Телефон: +998 91 234 56 78
    """
```

```python
masked_text = detector.mask_text(original_text)
print("Original Text:", original_text)
print("=="*20)
print("Masked Text:", masked_text)
```

```python
# Save the mapping for later use in Decryption
detector.save_mapping('pii_mapping.json')
print("\n✓ Saved PII mapping to 'pii_mapping.json'")
```

```python
detector.load_mapping('pii_mapping.json')
restored_text = detector.unmask_text(masked_text)
print("\nRestored Text:", restored_text)
```
